## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", color="green")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The colors are very vibrant and theres no color key either. The range of the graph looks like the difference in weight is significant and also the researcher could have used more species which might have been heavier or as heavy as the 'Gentoo' species.


How you changed the plot and why:  We changed the scale and started from 0 which lowers the exaguration of the scale.


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")

sns.scatterplot(data=diamonds, x="carat", y="price", color="blue", s=2)
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: The size of the dots on the graph were big for the data set and it was hard to read the graph but with a lower dot size the graph is much more readable.


How you changed the plot and why: changed the size of the dot from 100 to 2 which makes the graph more readable. This way its easier to tell where the market bulk lies.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, color="red", hue="year", palette="hsv")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: The unit was years which made the lines horizontal, we needed the data for the months in 1950s for which we needed the unit to be in months.


How you changed the plot and why: We changed the unit to months instead of years and to get all the data the aviation board needs they need to run this code with every year one by one, so that means they run it from 1950 to 1960 once every time.
